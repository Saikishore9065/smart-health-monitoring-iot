```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix,classification_report
from sklearn.preprocessing import StandardScaler
import seaborn as sns
```

```python
df = pd.read_csv("/content/Epics Dataset.csv")
```

```python
df
```

```python
x = df.iloc[:,:-1].values
```

```python
y=df.iloc[:,-1].values
```

```python
df.corr()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=0)
```

```python
sc = StandardScaler()
```

```python
x_train = sc.fit_transform(x_train)
x_test = sc.fit_transform(x_test)
```

SUPPORT VECTOR MACHINE

```python
SVC_model = SVC()
```

```python
SVC_model.fit(x_train, y_train)
```

```python
SVC_prediction = SVC_model.predict(x_test)
```

```python
print("Accuracy: {0:.2f}".format((accuracy_score(SVC_prediction, y_test)*100)))
```

NAVIE BAYES

```python
from sklearn.naive_bayes import GaussianNB
classifier = GaussianNB()
```

```python
classifier.fit(x_train, y_train)
```

```python
bay_prediction = classifier.predict(x_test)
```

```python
print("Accuracy: {0:.2f}".format((accuracy_score(bay_prediction, y_test)*100)))
```

Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics
```

```python
clf = DecisionTreeClassifier()
```

```python
clf = clf.fit(x_train,y_train)
```

```python
y_pred = clf.predict(x_test)
```

```python
print("Accuracy: {0:.2f}".format(metrics.accuracy_score(y_test, y_pred)*100))
```

Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
classifier= RandomForestClassifier(n_estimators= 10, criterion="entropy")
classifier.fit(x_train, y_train)
```

```python
RandomForestClassifier(criterion='entropy', n_estimators=10)
```

```python
y_pred= classifier.predict(x_test)
```

```python
print("Accuracy: {0:.2f}".format(accuracy_score(y_test, y_pred)*100))
```
